# NATEQ — Multi-Book RAG Quickstart

Select a registered book by ID. Each book has an isolated index and its own citations.

```python
from src.application.book_manager import BookManager

# Model objects are shared, but every book has its own Chroma index.
manager = BookManager(data_dir="data")
manager.list_books()
```

```python
# Ask the bundled book.
selected_book = "salah_al_din"
question = "أين انتقل صلاح الدين مع والده بعد مغادرة تكريت؟"

# This may download model weights and build the book index on first run.
result = manager.ask(question, book_id=selected_book)
print(result["answer_with_sources"])
```

```python
# Only the sources actually cited by the LLM are displayed.
for source in result["cited_sources"]:
    print(source["reference"], source["title"], source["page_number"])
    print(source["excerpt"][:300], "\n")
```
